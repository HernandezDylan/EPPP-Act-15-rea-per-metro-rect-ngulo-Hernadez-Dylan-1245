🟫 Cálculos de un rectángulo
🧾 Redacción del problema
Se requiere desarrollar un programa que permita calcular el área y el perímetro de un rectángulo.
El programa debe mostrar un menú con opciones para seleccionar el tipo de cálculo. El usuario ingresará la base y la altura del rectángulo, y el sistema realizará la operación correspondiente.
Se utilizarán estructuras de control como while para repetir el menú y if, elif, else para seleccionar la opción deseada.

🔹 Variables
| Variable  | Descripción                    | Tipo   |
|-----------|--------------------------------|--------|
| base      | Base del rectángulo            | Real   |
| altura    | Altura del rectángulo          | Real   |
| area      | Resultado del área             | Real   |
| perimetro | Resultado del perímetro        | Real   |
| opcion    | Opción seleccionada del menú   | Entero |


📐 Fórmulas
Área:
base × altura

Perímetro:
2 × (base + altura)

```python
# ╔══════════════════════════════════════════════╗
# ║        🟫 CÁLCULOS DE RECTÁNGULO 🟫          ║
# ║        Hernadez Dylan NL = 24                 ║
# ╚══════════════════════════════════════════════╝

# ════════════════ MENÚ ════════════════
def menu():
    print("\n╔════════════════════════════════════╗")
    print("║           📌 MENÚ PRINCIPAL        ║")
    print("╠════════════════════════════════════╣")
    print("║ 1️⃣  Área del rectángulo           ║")
    print("║ 2️⃣  Perímetro del rectángulo      ║")
    print("║ 3️⃣  Salir                         ║")
    print("╚════════════════════════════════════╝")


# ════════════════ ÁREA ════════════════
def arearectangulo():
    print("\n╔════════════════════════════════════╗")
    print("║        🟢 ÁREA DEL RECTÁNGULO      ║")
    print("╚════════════════════════════════════╝")

    base = float(input("🔹 Ingresa la base: "))
    altura = float(input("🔹 Ingresa la altura: "))

    area = base * altura

    print("\n╔════════ RESULTADO ════════╗")
    print(f"║ Base   : {base:<12} ║")
    print(f"║ Altura : {altura:<12} ║")
    print(f"║ Área   : {area:<12.2f} ║")
    print("╚══════════════════════════╝")


# ═══════════════ PERÍMETRO ═══════════════
def perimetrorectangulo():
    print("\n╔════════════════════════════════════╗")
    print("║     🔵 PERÍMETRO DEL RECTÁNGULO    ║")
    print("╚════════════════════════════════════╝")

    base = float(input("🔹 Ingresa la base: "))
    altura = float(input("🔹 Ingresa la altura: "))

    perimetro = 2 * (base + altura)

    print("\n╔════════ RESULTADO ════════╗")
    print(f"║ Base       : {base:<10} ║")
    print(f"║ Altura     : {altura:<10} ║")
    print(f"║ Perímetro  : {perimetro:<10.2f} ║")
    print("╚══════════════════════════╝")


# ═════════════ PROGRAMA PRINCIPAL ═════════════
while True:
    menu()
    opcion = int(input("👉 Selecciona una opción: "))

    if opcion == 1:
        arearectangulo()

    elif opcion == 2:
        perimetrorectangulo()

    elif opcion == 3:
        print("\n🚪 Cerrando programa...")
        break

    else:
        print("\n❌ Opción inválida, intenta de nuevo")

print("\n✔ Programa realizado por Hernadez Dylan NL=24")
```
